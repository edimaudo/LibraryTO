# Build the connected LibraryTO Relate graph

This notebook rebuilds the committed single-file Relate database from the seven CSV snapshots already included in the repository. It creates branch-to-observation, branch-to-event, branch-to-neighbourhood, branch-to-ward, branch-to-service and branch-to-facility relationships. It does not download or query any additional data source.

Run it from a local clone of `edimaudo/LibraryTO` with the application stopped. The script writes a staging `.rl` file, validates it, and only then replaces `data/tpl_library_graph.rl`.

```python
from pathlib import Path
import runpy

root = Path.cwd()
if (root / "scripts" / "build_relate_graph.py").is_file():
    repo_root = root
elif (root / ".." / "scripts" / "build_relate_graph.py").resolve().is_file():
    repo_root = (root / "..").resolve()
else:
    raise FileNotFoundError("Open this notebook from the repository root or its data/ directory. Clone the public LibraryTO repository first.")

runpy.run_path(str(repo_root / "scripts" / "build_relate_graph.py"), run_name="__main__")
```
